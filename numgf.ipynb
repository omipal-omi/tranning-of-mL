```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("imdevskp/corona-virus-report")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

print(os.listdir(path))  
```

```python
df = pd.read_csv(os.path.join(path, "covid_19_clean_complete.csv"))

# Kitne duplicate rows hain
print(df.duplicated().sum())

# Duplicate rows dekhne ke liye
print(df[df.duplicated(keep=False)])

# Duplicates hataane ke liye
df = df.drop_duplicates()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


latest = df[df["Date"] == df["Date"].max()]
state_data = latest.groupby("Province/State")["Confirmed"].sum().reset_index()


state_data = state_data[state_data["Province/State"] != "Unknown"]

plt.figure(figsize=(12, 5))
sns.barplot(
    data=state_data.sort_values("Confirmed", ascending=False).head(15),
    x="Province/State",
    y="Confirmed",
    #color ="pink"
   palette="viridis"

)
plt.title("Total COVID-19 Cases by State")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
rate_data = state_data.sort_values("Recover_Rate", ascending=False).head(10)

plt.figure(figsize=(12, 5))

sns.barplot(
    data=state_data.sort_values("Confirmed", ascending=False).head(15),
    x="Recover_Rate",
    y="Confirmed",
    hue="Province/State",
    palette="viridis",
    legend=False
)

plt.title("Total COVID-19 Cases by State")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```
